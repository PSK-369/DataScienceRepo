```python
mapping2={
    'Male':1,
    'Female':2
}

df2['Gender_encoded']=df2['Gender'].map(mapping2)
```

```python
le=LabelEncoder()

df2['Education_encoded']=le.fit_transform(df2['Education'])

mapping={
    "Suburban":1,
    "Rural":2,
    "Urban":3
}
df2['Location_encoded']=df2['Location'].map(mapping)
```

```python
df2.columns.to_list()
```

```python
df3=pd.DataFrame(data=df2,columns=['Age','Experience','Education_encoded', 'Location_encoded', 'Gender_encoded','Salary'],dtype='int')
df3.head(10)


#sort values

```

```python
import warnings
warnings.filterwarnings('ignore')

try:
    pass
    x=df3[['Age','Experience','Education_encoded','Location_encoded','Gender_encoded']]
    y=df3['Salary']
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.20,random_state=42)
    linear_reg=LinearRegression()
    linear_reg.fit(x_train,y_train)
    y_salary_predicated=linear_reg.predict(x_test)
    #print(str(y_salary_predicated))
except Exception as e:
    print('=============================================')
    print('errors found')
    print(f"{str(e)}\n")
    print("====================================")
    
else:
    print('no errors main function')

finally:
    print(" finally block is executed")
    
print("==========================================")
try:
    print(mean_absolute_error(y_test,y_salary_predicated))
    print(r2_score(y_test,y_salary_predicated))
    print(root_mean_squared_error(y_test,y_salary_predicated))
    print(linear_reg.score(x_test,y_salary_predicated))
    print(linear_reg.intercept_)
    print(linear_reg.coef_)
except Exception as e1:
    print('=============================================')
    print('errors found')
    print(f"{str(e1)}\n")
    print("====================================")
else:
    print('no errors main function')
    
```
