```python
def LinearSearch(l1):
    for i in l1:
        if i == target:
            print("element is found")
            break
LinearSearch(l1)
```

```python
def BinarySearch(l1):
    lft=0
    rht=len(l1)-1
    while lft <= rht:
        mid=lft+(rht-lft)//2
        if target == l1[mid]:
            print(l1[mid])
            break
        elif target > l1[mid]:
            lft=mid+1
        elif target < l1[mid]:
            rht = mid -1
            
BinarySearch(l1)


```

```python
l1=[1,2,3,4,5,6,7,8,900]
l2=[1,2,3,4,5,6,8,10,12]

target=10
```

```python
def InterPolationSearch(l1,target):
    lft1=0
    rht1=len(l1)-1
    
    while lft1 <= rht1 and target >=l1[lft1] and target <=l1[rht1]:
        if lft1 == rht1:
            if l1[lft1]==target:
                return 1
            else:
                return -1
                
        pos=lft1+((target-l1[lft1])//(l1[rht1]-l1[lft1]))*(rht1-lft1)
        if target == l1[pos]:
            return 1 
        elif target > l1[pos]:
            lft1=pos+1
        else:
            rht1=pos-1
    return -1
    
r1=InterPolationSearch(l1,target=1)     
print(r1)      
        
```

```python
def ternarySearch(l1,t1):
    left1=0
    right1= len(l1)-1
    while left1 <= right1:
        mid1=left1 +(right1-left1)//3
        mid2=right1-(right1-left1)//3
        if t1 == l1[mid1]:
            return mid1
        elif t1 == l1[mid2]:
            return mid2
        elif t1 < l1[mid1]:
            right1 = mid1-1
        elif t1 > l1[mid2]:
            left1 = mid2+1
        else:
            left1=mid1+1
            right1=mid2-1
    return -1
r2=ternarySearch(l1,t1=900)
print(r2)
            
```
