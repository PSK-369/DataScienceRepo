```python
import pandas as pd

class Anomolies_detection:
    def __init__(self, data):
        self.data = data
        
    def anomolies_function(self):
        # Using IQR method (more robust for small data)
        q1 = self.data.quantile(0.25)
        q3 = self.data.quantile(0.75)
        iqr = q3 - q1
        # Standard outlier bounds
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr
        anomolies1=[]
        for each_anomolie in self.data:
            if each_anomolie < lower_bound or each_anomolie > upper_bound:
               anomolies1.append(each_anomolie)
        return anomolies1

# Test with your data
data = {
    "ages": [1, 2, 3, 4, 5, 10, 11, 23, 78, 90],
    "marks": [100, 101, 102, 103, 104, 190, 240, 349, 10, 9]
}
df1 = pd.DataFrame(data)

for col_name in df1.columns:
    detector = Anomolies_detection(df1[col_name])
    results = detector.anomolies_function()
    if results:
        print(f"{col_name}: {results}")
        droping_data=df1.drop(index=list(results))
    else:
        print(f" {col_name}")
```
