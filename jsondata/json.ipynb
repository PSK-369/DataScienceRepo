```python
import json

data = [
    {
        "id": i,
        "name": f"User{i}",
        "email": f"user{i}@example.com",
        "age": 20 + (i % 50),
        "city": ["New York", "Los Angeles", "Chicago", "Houston", "Phoenix"][i % 5]
    }
    for i in range(1, 1001)
]

with open("sample_data.json", "w") as f:
    json.dump(data, f, indent=2)
```
